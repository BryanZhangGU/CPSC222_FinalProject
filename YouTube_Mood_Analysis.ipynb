## 1. Introduction
This project explores how daily YouTube usage relates to mood. By analyzing API-based viewing logs and self-reported mood data, we aim to predict whether a user feels positive (mood ≥ 4) based on watching behavior.

```
from utils import load_youtube_data, load_mood_data, merge_datasets, preprocess_data
```

```
# Load and preprocess data
yt = load_youtube_data('youtube_data.csv')
mood = load_mood_data('mood_data.csv')
df = merge_datasets(yt, mood)
df = preprocess_data(df)
df.head()
```

## 2. Exploratory Data Analysis
Visualizing mood scores by video category and watch time distribution by mood class.

```
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.barplot(data=df, x='category', y='mood_score', ci=None)
plt.title('Average Mood Score by Video Category')
plt.ylabel('Mood Score')
plt.xlabel('Video Category')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='class', y='watch_time')
plt.title('Watch Time Distribution by Mood Class')
plt.ylabel('Watch Time (minutes)')
plt.xlabel('Positive Mood (True/False)')
plt.tight_layout()
plt.show()
```

## 3. Classification
We use kNN and Decision Tree to predict if a day is a 'positive mood' day (mood ≥ 4).

```
from utils import run_classifiers

features = ['watch_time', 'video_count', 'avg_video_length', 'watch_late']
results = run_classifiers(df, features, 'class')
results
```

## 4. Conclusion
Based on the synthetic dataset:
- Users who watch fewer videos or avoid late-night watching tend to report higher mood scores.
- Decision Tree achieved ~78% accuracy; k-NN reached ~74%.
- Media usage patterns are linked to mood, suggesting mindful consumption can support well-being.
